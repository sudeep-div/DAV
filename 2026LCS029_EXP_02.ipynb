```python
import  pandas as pd
import numpy as np
```

```python
data ={
    "Name":['Amit','Bebo','Chetan','Amit','Divya','Ravi'],
    "Age":[25,np.nan,30,25,28,22],
    "Marks":[85,90,np.nan,85,70,95],
    "City":["'Mysuru",'Mysuru','Bengaluru','Mysuru',np.nan,'Delhi']
}
df=pd.DataFrame(data)
print("Dataset:")
print(df)
print("\nShape of dataset(r,c):",df.shape)
print("\nColumn names",df.columns.tolist())
print("\ndatatypes of each column:")
print(df.dtypes)
print("\nDataset info:")
df.info()
```

```python
print("Descriptive statistics (numerical columns):")
print(df.describe())
print("\nMean:\n",df.mean(numeric_only=True))
print("\nMEdian:\n",df.median(numeric_only=True))
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\nMissing Values in each column: ")
print(df.isnull().sum())
print("Total missing values in dataset:",df.isnull().sum().sum())
```

```python
print("\nDuplicate rows in datatset:")
print(df[df.duplicated()])
print("\nNumber of duplicates rows:",df.duplicated().sum())
print("\nEDA process completed successfully")
```
